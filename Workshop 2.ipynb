```python
from ultralytics import YOLO
import cv2
import matplotlib.pyplot as plt
from IPython.display import clear_output

model = YOLO("yolov8n.pt")

cap = cv2.VideoCapture(0)
cap.set(cv2.CAP_PROP_FRAME_WIDTH, 640)
cap.set(cv2.CAP_PROP_FRAME_HEIGHT, 480)

frame_count = 0
last_result = None

while True:

    ret, frame = cap.read()

    if not ret:
        break

    frame_count += 1

    # Run YOLO only every 2nd frame
    if frame_count % 2 == 0:
        results = model(
            frame,
            imgsz=320,
            conf=0.5,
            verbose=False
        )

        last_result = results[0].plot()

    # Show latest detection
    if last_result is not None:
        display_frame = last_result
    else:
        display_frame = frame

    display_frame = cv2.cvtColor(
        display_frame,
        cv2.COLOR_BGR2RGB
    )

    clear_output(wait=True)

    plt.figure(figsize=(8, 5))
    plt.imshow(display_frame)
    plt.axis("off")
    plt.show()

cap.release()
```
